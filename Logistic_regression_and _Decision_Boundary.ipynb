# Logistic Rgression and Decision Boundary

```python
import numpy as np 
%matplotlib widget
import matplotlib.pyplot as plt
from lab_utils_common  import plot_data, sigmoid, draw_vthresh
plt.style.use("./deeplearning.mplstyle")
```

```python
x = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2,2], [1, 2.5]])
y = np.array([0, 0,0, 1, 1, 1]).reshape(-1,1)
```

# Plot the data

```python
fig, ax = plt.subplots(1,1, figsize=(4,4))
plot_data(x, y, ax)
ax.axis([0,4,0,3.5])
ax.set_ylabel("$x_1$")
ax.set_xlabel("$x_0$")
plt.show()
```

```python
#Plot sigmoid(z) over a range of values from -10 to 10
z = np.arange(-10,11)
fig, ax = plt.subplots(1,1,figsize=(5,3))
#Plot z vs sigmoid(z)
ax.plot(z, sigmoid(z), c="b")
ax.set_title("Sigmoid Function")
ax.set_xlabel("z")
ax.set_ylabel("sigmoid(z)")
draw_vthresh(ax, 0.5)
plt.show()
```

# Plotting the decision boundary

```python
x0 = np.arange(0,6)
x1 = 3 - x0
fig, ax = plt.subplots(1,1, figsize=(4,4))
#Plot the decision boundary
ax.plot(x0, x1, c="k", label="Decision Boundary")
ax.axis([0,6,0,6])

#Fill the region below th eline
ax.fill_between(x0, x1, 6, color="tab:blue", alpha=0.2)

#Plot th eorignal data points
plot_data(x, y, ax)
ax.set_ylabel("$x_1$")
ax.set_xlabel("$x_0$")
plt.show()
```
